# Student Performance Analysis

## Objective

The objective of this project is to analyze student performance data and identify factors that may influence final grades.

## Research Questions

1. Does study time affect final grades?
2. Do absences affect academic performance?
3. Is there any difference in performance based on gender?

```python
import pandas as pd

df = pd.read_csv("student_data.csv")
df.head()
```

## Dataset Overview

First, we examine the dataset structure and check for missing values.

```python
df.info()
```

```python
df.describe()
```

### Observation

The dataset contains 395 student records and 33 attributes.

No missing values were found in the dataset, making it suitable for analysis without additional data cleaning.

# Research Question 1

Does study time affect final grades?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(df["studytime"], df["G3"])

plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")
plt.title("Study Time vs Final Grade")

plt.show()
```

### Observation

The scatter plot shows the relationship between study time and final grade. Since study time is grouped into four categories, it is difficult to clearly compare average performance.

```python
average_grades = df.groupby("studytime")["G3"].mean()

average_grades.plot(kind="bar")

plt.xlabel("Study Time")
plt.ylabel("Average Final Grade")
plt.title("Average Grade by Study Time")

plt.show()
```

### Analysis

Students in higher study-time categories generally achieve higher average grades compared to students in lower study-time categories.

### Conclusion

Study time appears to have a positive influence on academic performance, although the relationship is not perfectly linear.

# Research Question 2

Do absences affect final grades?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(df["absences"], df["G3"])

plt.xlabel("Absences")
plt.ylabel("Final Grade (G3)")
plt.title("Absences vs Final Grade")

plt.show()
```

### Analysis

Most students have relatively few absences. Students with extremely high absence counts generally do not achieve the highest grades, although several exceptions exist.

### Conclusion

There appears to be a weak negative relationship between absences and final grades.

# Research Question 3

Is there a difference in performance based on gender?

```python
average_grade_by_sex = df.groupby("sex")["G3"].mean()

average_grade_by_sex.plot(kind="bar")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.show()
```

### Analysis

The average final grade of male students is slightly higher than that of female students in this dataset.

### Conclusion

A small difference in average performance is observed between genders. Further analysis would be required to determine whether this difference is statistically significant.

# Final Conclusion

This analysis explored several factors that may influence student performance.

Key findings:

- Students who spend more time studying generally achieve higher grades.
- Higher absence counts may negatively affect academic performance.
- A small difference in average grades was observed between genders.

Overall, study habits and attendance appear to play important roles in student success.
